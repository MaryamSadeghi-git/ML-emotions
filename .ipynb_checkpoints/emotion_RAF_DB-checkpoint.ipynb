```python
import os
import cv2
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import VGG16
from tensorflow.keras.models import load_model
import matplotlib.pyplot as plt
```

```python

# ۱. تعریف مسیرها
IMAGE_DIR = "aligned"
LABEL_FILE = "list_patition_label.txt"

# ۲. خواندن فایل لیبل
df = pd.read_csv(LABEL_FILE, sep=r"\s+", header=None, names=["filename", "label"])

# ۳. فیلتر کردن ۴ کلاس (4: خوشحال, 5: غمگین, 6: عصبانی, 7: خنثی)
selected_classes = [4, 5, 6, 7]
df = df[df["label"].isin(selected_classes)].copy()

# ۴. تبدیل لیبل‌ها به 0 تا 3
mapping = {6: 0, 4: 1, 5: 2, 7: 3}
df["label"] = df["label"].map(mapping)

# ۵. جایگزینی .jpg با _aligned.jpg برای تطابق دقیق با فایل‌های پوشه
df["filename"] = df["filename"].str.replace(".jpg", "_aligned.jpg", regex=False)

# ۶. جداسازی واقعی Train و Test
train_df = df[df["filename"].str.startswith("train")].copy()
test_df = df[df["filename"].str.startswith("test")].copy()

print(f"تعداد داده‌های آموزش: {len(train_df)}")
print(f"تعداد داده‌های تست: {len(test_df)}")
```

```python
# ۷. تابع بارگذاری و نرمال‌سازی تصاویر
TARGET_SIZE = (100, 100)

def load_images_and_labels(data_df):
    images = []
    labels = []
    for _, row in data_df.iterrows():
        img_path = os.path.join(IMAGE_DIR, row["filename"])
        if os.path.exists(img_path):
            img = cv2.imread(img_path)
            if img is not None:
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                img = cv2.resize(img, TARGET_SIZE)
                images.append(img)
                labels.append(row["label"])
            
    X = np.array(images, dtype="float32") / 255.0
    y = np.array(labels, dtype="int32")
    return X, y

print("\nدر حال بارگذاری تصاویر آموزش...")
X_train_all, y_train_all = load_images_and_labels(train_df)

print("در حال بارگذاری تصاویر تست رسمی...")
X_test_official, y_test_official = load_images_and_labels(test_df)

print("\n--- وضعیت نهایی داده‌ها ---")
print(f"ابعاد X_train_all: {X_train_all.shape}")
print(f"ابعاد X_test_official: {X_test_official.shape}")
```

```python


# ۱. تفکیک داده‌ها به train و validation
X_train, X_val, y_train, y_val = train_test_split(
    X_train_all, 
    y_train_all, 
    test_size=0.20, 
    random_state=42, 
    stratify=y_train_all
)

print(f"تعداد داده‌های آموزش: {X_train.shape[0]}")
print(f"تعداد داده‌های ارزیابی: {X_val.shape[0]}")

# ۲. بارگذاری مدل VGG16 از پیش آموزش‌دیده (بدون لایه‌های طبقه‌بندی آخر)
base_model = VGG16(
    weights='imagenet', 
    include_top=False, 
    input_shape=(100, 100, 3)
)

# فریز کردن وزن‌های VGG16 (تا وزن‌های قبلی تغییر نکنند)
base_model.trainable = False

# ۳. اضافه کردن لایه‌های سفارشی خودمان در انتهای مدل
model = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5), # جلوگیری از Overfitting
    layers.Dense(4, activation='softmax') # ۴ کلاس خروجی
])

# ۴. کامپایل مدل
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# ۵. آموزش مدل
EPOCHS = 10
BATCH_SIZE = 32

print("\nدر حال شروع آموزش مدل VGG16...")
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE
)

```

```python
model.save("raf_vgg16.keras")
```

```python


loaded_model = load_model("raf_vgg16.keras")
loaded_model.summary()
```

```python


# رسم نمودار Accuracy
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Accuracy Model')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

```python
# ارزیابی نهایی
loss, accuracy = model.evaluate(X_val, y_val)

print(f"دقت نهایی مدل روی داده‌های تست: {accuracy * 100:.2f}%")
```
